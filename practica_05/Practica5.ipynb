| Campo | Descripción |
|---|---|
| **#** | El ID o número de identificación del Pokémon. |

| **Type 1** | Se refiere a las propiedades elementales (tipos) asociadas tanto al Pokémon como a sus movimientos. |
| **Type 2** | Los Pokémon pueden tener hasta dos tipos (Pokémon de tipo Dual); los movimientos solo pueden tener uno. |
| **Total** | Representa la suma de todos los stats, lo que te da una idea de qué tan "pesado" o fuerte es el Pokémon. |
| **HP** | (Puntos de Salud) Es qué tanto aguante tiene un Pokémon antes de que lo dejen fuera de combate (se desmaye). |
| **Attack** | La fuerza de los ataques físicos del Pokémon. |
| **Defense** | Qué tanto aguanta el Pokémon contra los ataques físicos del rival. |
| **Evolution_Level** | Nivel de Evolución del Pokémon. |
| **Sp. Atk** | Ataque Especial: el poder de los movimientos especiales del Pokémon. |
| **Sp. Def** | Defensa Especial: Qué tan resistente es el Pokémon ante los ataques especiales. |
| **Speed** | Velocidad; el Pokémon que tenga más velocidad es el que ataca primero. |
| **Generation** | Generación: Se refiere a la generación de la serie de juegos a la que pertenece. |
| **Legendary** | Indica si el Pokémon es Legendario o no (son bichos muy raros y poderosos). Es un valor booleano (sí/no). |
| **Evolution_Level** | Indica la etapa evolutiva en la que se encuentra el Pokemon. Es un valor entre 1 y 4. |

26. Utilizar graficas de enjambre (SwarnPlots) para representar las estadisticas especificas por tipo de pokemon.

```python
n_cols = 3
n_rows = math.ceil(len(list_types) / n_cols)

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize = (17, 4 * n_rows),
    sharey = True
)

axes = axes.flatten()

for ax, tipo in zip (axes, list_types):
    data_tipo = poke_melt[poke_melt['Type'] == tipo]
    sns.stripplot(
    data = data_tipo,
    x = 'variable',
    y = 'value',
    hue = 'variable',
    palette = 'gist_stern',
    jitter = True,
    size = 3,
    alpha = 0.7,
    legend = False,
    ax = ax
    )

    ax.set_title(tipo)
    ax.set_xlabel('')
    ax.set_ylabel('Valor')
    ax.tick_params(axis = 'x', rotation = 45)

for ax in axes[len(list_types):]:
    ax.remove()

plt.tight_layout()
plt.show()
```

```python
totales_mean = (
    mis_pokemons
    .groupby('Type')
    .mean(numeric_only=True)   
    .Total
    .sort_values(ascending=False)
)

plt.figure(figsize=(15, 5))
sns.barplot(
    x=totales_mean.index,
    y=totales_mean.values,
    hue=totales_mean.index,   
    palette='twilight_shifted',
    legend=False,
    dodge=False,
)
plt.title('Mean of the total of all stats for each type of Pokémon')
plt.tight_layout()
plt.show()
```
