15. Conteo de pokemones por atributo **Tipo** (Graficado).
    Se usa la libreria matplot para configurar el tamaño de la grafica. 

```python
plt.figure(figsize=(10,5))
ax = sns.countplot(
    data=pokemons,
    x='Type',
    hue='Type',
    palette='twilight'
)
#Agregar totales en la parte superior de cada barra
for container in ax.containers:
    ax.bar_label(
        container,
        label_type='edge',
        padding=3
    )
plt.title('Número de pokemones agrupados por Tipo')
plt.xlabel('Tipo')
plt.ylabel('Cantidad')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

22. Calculando los datos para graficacion del total de estadisticas por cada tipo de pokemon, respetando el orden alfavetico previamnete definido.   

```python
stats = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']
palettes = ['magma','ocean', 'vlag', 'copper', 'mako', 'winter']

pokemon_stast = pokemons.groupby('Type').sum()[stats]

plt.figure(figsize=(17,17))

for i, stat in enumerate(stats, start=1):
    plt.subplot(3, 2, i)
    sns.barplot(
        x=pokemon_stast[stat], 
        y=pokemon_stast[stat], 
        palette=palettes[i-1]
        )
    plt.title(f'Suma total de {stat} por Tipo')
    plt.xlabel('stat')
    plt.ylabel('Type')
    plt.xticks(rotation=45) 
plt.tight_layout()
plt.show()
```

**Ejercicio 13:** Recalcular los graficos de Estadisticas para el 2 dataframe

```python
stats = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']
palettes = ['magma','ocean', 'vlag', 'copper', 'mako', 'winter']

pokemon_stast = mis_pokemons.groupby('Type').sum()[stats]

plt.figure(figsize=(17,17))

for i, stat in enumerate(stats, start=1):
    plt.subplot(3, 2, i)
    sns.barplot(
        x=pokemon_stast[stat], 
        y=pokemon_stast[stat], 
        palette=palettes[i-1]
        )
    plt.title(f'Suma total de {stat} por Tipo')
    plt.xlabel('stat')
    plt.ylabel('Type')
    plt.xticks(rotation=45) 
plt.tight_layout()
plt.show()
```

23. Grafica de colmena (Swarmplots)

```python
k = 1

plt.figure(figsize=(15,30))

for stat in stats:
    plt.subplot(6, 1, k)

    sns.stripplot(
        data=pokemons,
        x='Type',
        y=stat,
        hue='Type',
        palette='Dark2',
        legend=False
    )

    plt.title(f'Total of {stat} for each Type')
    plt.xlabel('Type')
    plt.ylabel(stat)
    plt.xticks(rotation=45)

    k += 1

plt.tight_layout()
plt.show()
```
